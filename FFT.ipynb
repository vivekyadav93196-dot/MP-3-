```python
# COMPUTATION OF DFT USING COMPLEX NUMBERS ---
import numpy as np

def dft(x):
    N = len(x)
    X = np.zeros(N,dtype=complex)          # for storing dft coefficients

    for k in range(N):
        for n in range(N):
            angle = (2*np.pi*k*n)/N
            X[k] += x[n]*np.exp(-1j * angle)
            return X

x = np.array([1,2,3,4])
X = dft(x)

print("Input Signal:",x)
print("\nDFT Coefficients:")

for k in range(len(X)):
    print(f'X({k}) = {X[k].real:.2f} + {X[k].imag:.2f}')     
```
