```python
# CALCULATING THE MODULUS OF COMPLEX NUMBERS ---

real_p = float(input("Enter the real part:"))
imgry_p = float(input("Enter the imaginary part:"))    

z = complex(real_p,imgry_p)

mod = math.sqrt(z.real**2 + z.imag**2)

print("Complex number =", z)
print("Modulus =", mod)
```

```python
# CALCULATING THE CONJUGATE OF COMPLEX NUMBER ---

z = 3 + 6j
con = z.conjugate()

print(z)
print("Conjugate:", con)
```

```python
# COMPLEX NUMBER ARITHMETIC ---
z1 = complex(input("Enter 1st complex number :"))
z2 = complex(input("Enter 2nd complex number :"))

print("\nResults")
print("Addition :", z1+z2)
print("Subtraction :", z1-z2)
print("Multiplication :", z1*z2)
print("Division :", z1/z2)
```

```python
# PLOTTING OF COMPLEX NUMBERS AS ORDERED PAIRS ---
import matplotlib.pyplot as plt
z = 2 + 9j

x = z.real
y = z.imag

plt.plot(x,y,'o')
plt.xlabel('Real Axis')
plt.ylabel('Imaginary Axis')
plt.title("Complex number as Ordered Pair")

plt.text(x,y,f"({x},{y})")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
z = [2+3j, -1+6j, -6+2j, 8+4j]

for num in z:
    plt.plot(num.real, num.imag, 'o')
    plt.text(num.real, num.imag, f'{num}')

plt.axhline(0)
plt.axvline(0)

plt.xlabel('Real Axis')
plt.ylabel('Imaginary Axis')
plt.title("Complex number on the complex plane")

plt.grid(True)
plt.show()
```

```python
# CONVERSION OF COMPLEX NUMBER FROM CARTESIAN TO POLAR ---
import math 
a = float(input("Enter the real part:"))
b = float(input("Enter the imaginary part:")) 

r = math.sqrt(a**2 + b**2)
theta = math.atan2(b,a)

theta_deg = math.degrees(theta)

print("Modulus=", r)
print("Argument=", theta_deg)
print('Polar form=', r, "(cos(",theta_deg,")+ isin(",theta_deg,"))")
```
