```python
# Determining the nth roots of a complex number and repressents in cartesian and polar form ----
import math

a = float(input("Enter real part::"))
b = float(input("Enter imaginary part::"))

n = int(input("Enter the value of n:"))

r = math.sqrt(a**2 + b**2)
theta = math.atan2(b,a)

root_n = r**(1/n)
for k in range(n):
    angle = (theta + 2*math.pi*k)/n
    real = root_n*math.cos(angle)
    imgry = root_n*math.sin(angle)

    print("\nRoot", k+1)
    print("Cartesian form:")
    print(real,"+",imgry,'i')
    print("Polar form:")
    #print(root_n,"(cos(",angle,")+isin(",angle,"))")
    print(f"{root_n:.2f}(cos({angle:.2f}) + isin({angle:.2f}))")  
```

```python
# TRANSFORMATION OF COMPLEX NUMBERS AS 2-D VECTORS ---
import cmath
import math

z = complex(input("Enter complex number:"))
print("Original complex number:", z)

#Translation--
translation = 1 + 2j
z_translation = z + translation
print("\nAfter translation:",z_translation)

#Scaling--
scale = 2
z_scaling = scale*z
print("After scaling:", z_scaling)

#Rotation--
angle = 45
theta = math.radians(angle)
z_rotation = z*cmath.exp(1j*theta)
print("After rotation:",z_rotation)

#Reflection--
z_reflection= z.conjugate()
print("After reflection:",z_reflection)
```

```python
# VISUALIZATION OF MAPPPINGS OF SOME ELEMENTARY COMPLEX FUNCTIONS w = f(Z) FROM Z-PLANE TO W-PLANE ----

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2,2,15)
y = np.linspace(-2,2,15)

X,Y = np.meshgrid(x,y)
z = X+1j*Y

w = z**2
U = w.real
V = w.imag

plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.scatter(X,Y)
plt.title('z-plane')
plt.xlabel("Real axis")
plt.ylabel("Imaginary axis")
plt.grid()

plt.subplot(1,2,2)
plt.scatter(U,V)
plt.title('w-plane: w = z**2')
plt.xlabel("Real axis")
plt.ylabel("Imaginary axis")
plt.grid()

plt.tight_layout()
plt.show()
```
