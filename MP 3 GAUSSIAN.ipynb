```python
# SOLVING A GAUSS LEGENDRE POLYNOMIAL ---

import numpy as np
def f(x):
    return x**2 + 2*x + 1

n = 3   # no. of gauss legendre points

x,w = np.polynomial.legendre.leggauss(n)
a=int(input("Enter limit a:"))
b=int(input("Enter limit b:"))

integral = 0
for i in range(n):
    X = ((b-a)*x[i]+(b+a))/2
    integral = integral + w[i]*f(X)
    
integral = ((b-a)/2)*integral
print("Value of integral =",integral)
```

```python
# REPRESENTATION OF A FUNCTION USING LEGENDRE POLYNOMIALS ---
import numpy as np

def f(x):
    return x**2 + x + 1

def P0(x):
    return 1    
def P1(x):    
    return x
def P2(x):
    return (3*x**2 - 1)/2

n = 5            # numner of gaussian points

x,w = np.polynomial.legendre.leggauss(n)

a0 = 0
for i in range(n):
    a0 = a0 + w[i]*f(x[i])*P0(x[i])
a0 = (1/2)*a0

a1 = 0
for i in range(n):
    a1 = a1 + w[i]*f(x[i])*P1(x[i])
a1 = (3/2)*a1

a2 = 0
for i in range(n):
    a2 = a2 + w[i]*f(x[i])*P2(x[i])
a2 = (5/2)*a2

print("a0=", a0)
print("a1=", a1)
print("a2=", a2)

print("\nRepresentation of function:")
print("f(x)=", round(a0,3), "P0(x) +", round(a1,3), "P1(x) +", round(a2,3), "P2(x)")
```

```python
# SOLVING IMPROPER INTEGRALS USING GAUSS LAGAUERRE ---

import numpy as np

def f(x):
    return 1/(1+x**2)
def gauss_laguerre(n):
    x,w = np.polynomial.laguerre.laggauss(n)
    
    I = 0
    for i in range(n):
        I = I + w[i]*f(x[i])
    return I

n = 10
I = gauss_laguerre(n)
print("Gauss-Laguerre result=", round(I,8))
```

```python
# SOLVING IMPROPER INTEGRALS USING GAUSS-HERMITE QUADRATURE ---

import numpy as np

def f(x):
    return 1

def gauss_hermite(n):
    x,w = np.polynomial.hermite.hermgauss(n)

    I = 0
    for i in range(n):
        I = I + w[i]*f(x[i])
    return I

n = 10
I = gauss_hermite(n)
exact = np.sqrt(np.pi)
error = abs(exact - I)

print("Gauss-Hermite result=", round(I,8))
print("Analytical result=", round(exact,8))
print("Absolute error=",format(error,".2e"))
```
